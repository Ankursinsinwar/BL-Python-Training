* Using Pandas, we can accomplish five typical steps in the processing and analysis of data, regardless of the origin of data — load, prepare, manipulate, model, and analyze.

## DataFrame
A Data frame is a two-dimensional data structure, i.e., data is aligned in a tabular fashion in rows and columns

### Create an Empty DataFrame

```python
# A basic DataFrame, which can be created is an Empty Dataframe.
import pandas as pd
df = pd.DataFrame()
print(df)
```

### Create a DataFrame from Dict of lists/ndArray
* All the ndarrays must be of same length.
* If index is passed, then the length of the index should equal to the length of the arrays.
* If no index is passed, then by default, index will be range(n), where n is the array length.

```python
# Ex1) without index
name = ['Ashay', 'Himanshu', 'Padam', 'Ravi']
age = [22,23,24,25]

# creating dictionary
data = {'Name':name,'Age':age}
print(data)
# passing dictionary to create df
df = pd.DataFrame(data)
df
```

```python
#Ex2) with index
name = ['Ashay', 'Himanshu', 'Padam', 'Ravi']
age = [22,23,24,25]

# my new indexes
indexes = ['rank1','rank2','rank3','rank4']

#this is my dictionary
data = {'Name':name,'Age':age}

#passing dictionary to create dataframe
df = pd.DataFrame(data, index=indexes)
df
```

### Column Selection

```python
df['Age']
```

```python
df.Age
```

### Column Addition

```python
# say suppose you want to add city column
city = ['Banglore','Pune','Hyderabad','Mumbai']

#adding new column named 'City'
df["City"] = city
df
```

### Column Deletion

```python
# 1) Using del keyword
del df["City"]
df
```

```python
# 2) Using pop function
df.pop('Age')
df
```

##  Data cleaning

```python
# creating a dataframe
dict1 = {
    0:[1,4,7,np.nan],
    1:[2,5,np.nan,np.nan],
    2:[3,np.nan,np.nan,np.nan],
    3:[np.nan,np.nan,np.nan,np.nan]
}

df = pd.DataFrame(dict1)
df
```

```python
#to detect missing values
df.isnull()
```

```python
# Detect existing (non-missing) values.
df.notnull()
```

```python
#drops an entire row if one nan value is present
df.dropna()
```

```python
# deletes the rows in which all the values are null
df.dropna(how='all')
```

```python
# deletes the columns in which all the values are null
df.dropna(axis=1, how='all')
```

```python
# threshold Require that many 'non-nan' values.
df.dropna(thresh=1)
```

```python
df.dropna(thresh=1,axis = 1)
```

```python
df
```

```python
# to fill perticular values in all NaN
df_fill = df.fillna(10)
df_fill
```

```python
# to fill perticular values in perticular columns
df_fill = df.fillna({0: 10, 1: 20, 2: 30, 3: 40})
df_fill
```

```python
# fills the nan value with previous observation/row value
df_fill = df.fillna(value='NA')
df_fill
```

```python
#if you specify the limit,then only for those many 'NaN' values in a column it will fill the values
df_fill = df.fillna(value ={0: 10, 1: 20, 2: 30, 3: 40}, limit = 2)
df_fill
```

```python
# fills the nan value with previous mean of column
df_mean = df.fillna(df.mean())
df_mean
```

```python
# fills the nan value with previous mean of row
df_mean = df.fillna(df.mean(), axis=1)
df_mean
```

```python
# fills the nan value with previous median of column
df_median = df.fillna(df.median())
df_median
```
