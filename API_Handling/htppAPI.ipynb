```python
import requests
```

```python
url = "http://localhost:3000/data"
# https://jsonplaceholder.typicode.com/
```

```python
data = {
    "name": "Ankur",
    "age": 30
}

response = requests.post(url, json=data)

print(response.status_code)
print(response.json())
```

```python

response = requests.get(url)

print(response.status_code)
print(response.json())
```

```python
data = {
    "name": "Amit",
    "age": 20
}

response = requests.post(url, json=data)

print(response.status_code)
print(response.json())
```

```python
puturl = "http://localhost:3000/data/kBwfC9UkNLs"

updated_data = {
    "name": "Ankur Updated",
    "age": 23
}

response = requests.put(puturl, json=updated_data)

print(response.status_code)
print(response.json())
```

```python
response = requests.patch(puturl, json={"age": 24})

print(response.json())
```

```python
import requests

url = "http://localhost:3000/data"

headers = {
    "Authorization": "Bearer YOUR_API_KEY",
    "Content-Type": "application/json"
}

response = requests.get(url, headers=headers)

print(response.json())
```
