```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

# **Line Graph**

```python
price = [5459,4543,4558,1414]
years = [5,6,7,5]
plt.plot(years,price,'g*',linestyle = "solid")
plt.title("P/Y")
plt.show()
```

```python
x = [1, 2, 3]
y1 = [1, 2, 3]
y2 = [3, 2, 1]

plt.plot(x, y1, label='Line 1')
plt.plot(x, y2, label='Line 2')

plt.legend()
plt.show()
```

# **Scatter graph**

```python
x= np.linspace(-10,10,50)
y = 10*x+3+np.random.randint(0,200,50)
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.scatter(x,y)
```

# **Pie graph**

```python
data = [23,45,100,20,49]
sub = ['eng','science','math','sst','hindi']
plt.pie(data,labels=sub,autopct='%0.1f%%',explode=[0,0,.1,0,0])
plt.legend()
plt.show()
```

# **Bar gaph**

```python
Y=['a','b','c','d']
X=[2,3, 5, 4]
plt.bar(Y,X,color=['red','green','blue','black'])
```


# **Histogram gaph**

```python
data = np.random.randn(1000)

plt.hist(data, 
         bins=10,
         density=True, 
         color='green', 
         edgecolor='black')    
```
